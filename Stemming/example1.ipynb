PORTER STEMMER

```python
words = ["eating","eats","eat","writing","write","writes","go","goes","going"]
```

```python
from nltk.stem import PorterStemmer
```

```python
Stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"---"+ Stemming.stem(word))
```

```python
Stemming.stem("congratulations") # here the meaning is changed , so this is the disadvantage. 
```

RegexStemmer class

```python
from nltk.stem import RegexpStemmer
```

```python
regStemmer = RegexpStemmer('ing$|s$|e$|able$' , min=4)
```

```python
regStemmer.stem('eating')
```

```python
regStemmer.stem('ingeating')
```

Snowball Stemmer - better accuracy - better form of word

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word + "---" + snowballstemmer.stem(word))
```
